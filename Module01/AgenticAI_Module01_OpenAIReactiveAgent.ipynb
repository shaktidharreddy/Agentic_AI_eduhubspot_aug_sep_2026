# Step 1 - Import required modules

* OpenAI() is the official client interface
* userdata.get() reads secrets stored in Colab
* os.environ supports environment-variable setup
* Keeps keys out of notebook code for safety

```python
from openai import OpenAI
from google.colab import userdata
import os
```

# Step 2 - Load API key from Colab Secrets

* Prevents hardcoding secrets in notebooks
* Matches recommended API key handling (env variable)
* Fails fast with a clear error if missing
* Creates a reusable client for all later calls

```python
api_key = userdata.get('Eduhubspot_Key23')

if not api_key:
    raise ValueError("API key not found. Add 'Eduhubspot_Key23' in Colab Secrets.")

os.environ["OPENAI_API_KEY"] = api_key  # SDK reads from env automatically
client = OpenAI()
print("OpenAI client configured successfully!")
```

# Step 3 - Create a simple reactive chatbot function

* Single-turn interaction = reactive behavior
* System message controls tone and style
* User message contains the learner question
* Returns only the final generated text

```python
def reactive_chatbot(user_text: str) -> str:
    response = client.responses.create(
        model="gpt-4.1-mini",
        input=[
            {"role": "system", "content": "You are a helpful training assistant. Keep answers short and clear."},
            {"role": "user", "content": user_text}
        ],
    )
    return response.output_text
```

# Step 4 - Test the chatbot (sample inputs)

* Confirms API connectivity works
* Confirms function returns text output
* Validates instruction-following (short & clear)
* Demonstrates typical learner questions

```python
print(reactive_chatbot("Explain Agentic AI in 3 bullet points."))
print()
print(reactive_chatbot("What is the difference between an AI agent and a chatbot?"))
```

# Step 5 - Run a small chat loop

* Supports multi-turn chatting without memory
* Each response is still independent (reactive)
* exit/quit cleanly ends the session

```python
while True:
    user_text = input("You: ")
    if user_text.strip().lower() in {"exit", "quit"}:
        print("Chat ended.")
        break
    bot_text = reactive_chatbot(user_text)
    print("Bot:", bot_text)
```
