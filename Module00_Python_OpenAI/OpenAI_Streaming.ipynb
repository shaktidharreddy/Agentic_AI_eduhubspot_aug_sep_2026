```python
from openai import OpenAI

client = OpenAI()

stream = client.responses.create(
    model="gpt-6-luna",  # or another reasoning model like o3, gpt-5, etc.
    input="How many r's are in strawberry? Think step by step.",
    reasoning={"summary": "auto"},  # Required to emit reasoning summary events
    stream=True,
)

for event in stream:
    # 1. Internal reasoning summary chunks
    if event.type == "response.reasoning_summary_text.delta":
        print(event.delta, end="", flush=True)

    # Optional: Separate reasoning from the final answer when reasoning finishes
    elif event.type == "response.reasoning_summary_text.done":
        print("\n\n--- Final Answer ---\n")

    # 2. Final user-facing answer chunks
    elif event.type == "response.output_text.delta":
        print(event.delta, end="", flush=True)
```

```python
from openai import OpenAI

# Requires OPENAI_API_KEY in your environment variables
client = OpenAI()

stream = client.responses.create(
    model="gpt-6-luna", # Works with any reasoning model (o3, o4-mini, gpt-5, etc.)
    input="How many r's are in strawberry? Think step by step.",
    reasoning={"summary": "auto"},  # Enables reasoning summary events
    stream=True,
)

in_reasoning = False

for event in stream:
    # 1. Stream internal reasoning thoughts first
    if event.type in (
        "response.reasoning_summary_text.delta",
        "response.reasoning_text.delta",
    ):
        if not in_reasoning:
            print("=== Reasoning ===\n", flush=True)
            in_reasoning = True
        print(event.delta, end="", flush=True)

    # 2. Close out the reasoning block when done
    elif event.type in (
        "response.reasoning_summary_text.done",
        "response.reasoning_text.done",
    ):
        if in_reasoning:
            print("\n\n=== Final Answer ===\n", flush=True)
            in_reasoning = False

    # 3. Stream the final answer output
    elif event.type == "response.output_text.delta":
        print(event.delta, end="", flush=True)

    # 4. Handle stream errors
    elif event.type == "error":
        print(f"\n[Stream Error]: {event.error}")

print()  # Final newline
```

```python
from openai import OpenAI

client = OpenAI()

stream = client.responses.create(
    model="gpt-6-luna",  # Or "gpt-6.2-sol" / "gpt-6.2-astra"
    input="How many r's are in strawberry? Think step by step.",
    reasoning={"effort": "medium", "summary": "auto"},
    stream=True,
)

in_reasoning = False

for event in stream:
    # 1. Stream internal reasoning thoughts first
    if event.type in (
        "response.reasoning_summary_text.delta",
        "response.reasoning_text.delta",
    ):
        if not in_reasoning:
            print("=== Reasoning ===\n", flush=True)
            in_reasoning = True
        print(event.delta, end="", flush=True)

    # 2. Separate reasoning from the final answer
    elif event.type in (
        "response.reasoning_summary_text.done",
        "response.reasoning_text.done",
    ):
        if in_reasoning:
            print("\n\n=== Final Answer ===\n", flush=True)
            in_reasoning = False

    # 3. Stream the final answer output
    elif event.type == "response.output_text.delta":
        print(event.delta, end="", flush=True)

    elif event.type == "error":
        print(f"\n[Stream Error]: {event.error}")

print()
```
